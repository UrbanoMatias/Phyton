**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
# evaluate the contents of the file
from os.path import exists
import csv

# Load the customers file
datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Read customers into a list of dictionaries
customers = []

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # Read header row

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            customer = {header[i].strip(): row[i].strip() for i in range(len(header))}
            customers.append(customer)

print(f"✅ Total customers loaded: {len(customers)}")

# Create a set to store unique states
unique_states = set()

for customer in customers:
    state = customer.get("State", "").strip().upper()
    if state:
        unique_states.add(state)

print(f"🌎 Number of different states where customers live: {len(unique_states)}")
print(f"States: {sorted(unique_states)}")

```

**Question 3** What is the state with most customers?

```python
# open file and init vars
# evaluate the contents of the file
from collections import defaultdict

# Count customers per state
state_counts = defaultdict(int)

for customer in customers:
    state = customer.get("State", "").strip().upper()
    if state:
        state_counts[state] += 1

# Find the state with the highest customer count
most_common_state = max(state_counts, key=state_counts.get)
count = state_counts[most_common_state]

print(f"🏆 State with the most customers: {most_common_state} ({count} customers)")

```

**Question 4** What is the state with the least customers?

```python
# Find the minimum customer count
min_count = min(state_counts.values())

# Find all states with that minimum count
least_states = [state for state, count in state_counts.items() if count == min_count]

print(f"📉 States with the least customers ({min_count} each): {', '.join(sorted(least_states))}")
```

**Question 5:** What is the most common last name?

```python
# open file and init vars
# evaluate the contents of the file
# Count last names
last_name_counts = defaultdict(int)

for customer in customers:
    last_name = customer.get("LastName", "").strip().lower()
    if last_name:
        last_name_counts[last_name] += 1

# Find the most frequent last name
most_common_lastname = max(last_name_counts, key=last_name_counts.get)
count = last_name_counts[most_common_lastname]

print(f"👨‍👩‍👧 Most common last name: {most_common_lastname.capitalize()} ({count} customers)")
```

```python
# open and evaluate orders.csv
# open file and init vars
# evaluate the contents of the file
from os.path import exists
import csv
from collections import defaultdict

# Load and evaluate orders.csv
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Question 0: Count total rows (not unique orders)
with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # Read header

    total_rows = 0
    for row in csvreader:
        # Skip empty or malformed rows
        if row and any(cell.strip() for cell in row):
            total_rows += 1

    print(f"Total rows in orders.csv: {total_rows}")
print("All done!")
```

```python
# Question 1: How many unique orders are in the orders.csv file?
unique_orders = set()

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)
    order_id_index = header.index("OrderID")

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            order_id = row[order_id_index].strip()
            if order_id:
                unique_orders.add(order_id)

print(f"📦 Unique orders found: {len(unique_orders)}")

```

```python
# Question 2: What is the average number of items per order?
order_items = defaultdict(int)

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)
    order_id_index = header.index("OrderID")

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            order_id = row[order_id_index].strip()
            order_items[order_id] += 1

total_orders = len(order_items)
total_items = sum(order_items.values())
average_items = round(total_items / total_orders, 2)

print(f"📊 Average items per order: {average_items}")

```

```python
# Question 3: What is the highest number of items per order?
max_items = max(order_items.values())
# Find the order(s) with that max item count
largest_orders = [oid for oid, count in order_items.items() if count == max_items]

print(f"🔝 Highest item count in a single order: {max_items}")
print(f"Order(s) ID with {max_items} items: {', '.join(largest_orders)}")

```

```python
# Question 4: How many orders were placed in October 2021?
october_orders = set()

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)
    order_id_index = header.index("OrderID")
    date_index = header.index("Date")

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            date = row[date_index].strip()
            order_id = row[order_id_index].strip()
            if date.startswith("2021-10"):
                october_orders.add(order_id)

print(f"📅 Orders placed in October 2021: {len(october_orders)}")

```

```python
from collections import defaultdict
import csv

# Load the orders file
orders_file = "./sample_data/orders.csv"
customers_file = "./sample_data/customers.csv"

# Dictionary to store total per order (only once)
order_totals = {}
order_customers = {}

with open(orders_file, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)

    # Get column indexes
    customer_index = header.index("CustomerID")
    date_index = header.index("Date")
    order_id_index = header.index("OrderID")
    total_index = header.index("OrderTotal")

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            date = row[date_index].strip()
            if date.startswith("2021-"):
                order_id = row[order_id_index].strip()
                customer_id = row[customer_index].strip()
                # Only count OrderTotal once per order
                if order_id not in order_totals:
                    try:
                        total = float(row[total_index].strip())
                        order_totals[order_id] = total
                        order_customers[order_id] = customer_id
                    except ValueError:
                        continue  # Skip invalid totals

# Sum totals per customer
customer_spending = defaultdict(float)
for order_id, total in order_totals.items():
    customer_id = order_customers[order_id]
    customer_spending[customer_id] += total

# Find top spender
top_customer = max(customer_spending, key=customer_spending.get)
top_amount = round(customer_spending[top_customer], 2)

# Load customer names from customers.csv
customer_names = {}

with open(customers_file, 'r') as fl:
    csvreader = csv.reader(fl)
    header = next(csvreader)

    id_index = header.index("CustomerID")
    first_index = header.index("FirstName")
    last_index = header.index("LastName")

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            cid = row[id_index].strip()
            name = f"{row[first_index].strip()} {row[last_index].strip()}"
            customer_names[cid] = name

# Format final output
top_name = customer_names.get(top_customer, "Unknown")
print(f"{top_name}, CustomerId: {top_customer} has spent ${top_amount}")
```

```python
# Question 6: What is the best month for sales historically?
from collections import defaultdict
from os.path import exists
import csv

# Dictionary to accumulate sales per month (format: YYYY-MM)
monthly_sales = defaultdict(float)

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)

    # Get column indexes
    date_index = header.index("Date")
    total_index = header.index("OrderTotal")

    for row in csvreader:
        if row and any(cell.strip() for cell in row):
            date = row[date_index].strip()
            # Ensure date has at least YYYY-MM format
            if len(date) >= 7:
                month = date[:7]  # Extract YYYY-MM
                try:
                    total = float(row[total_index].strip())
                    monthly_sales[month] += total
                except ValueError:
                    continue  # Skip rows with invalid totals

# Find the month with the highest total sales
best_month = max(monthly_sales, key=monthly_sales.get)
best_total = round(monthly_sales[best_month], 2)

print(f"🏆 Best sales month: {best_month} with total sales of ${best_total}")
```
